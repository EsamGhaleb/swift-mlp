### 1.1 Quick Environment Check
Before you get started, we recommended running this cell to ensure all dependencies are met.

````python
import sys
import shutil
import subprocess


def check_env():
    print("--- Environment Verification ---")
    print(f"Python: {sys.version.split(' ')[0]}")
    print(f"Path:   {sys.executable}")
    print("-" * 30)
    
    # 1. FFmpeg Check
    ffmpeg_path = shutil.which("ffmpeg")
    if ffmpeg_path:
        print(f"✅ FFmpeg: FOUND at {ffmpeg_path}")
        try:
            res = subprocess.run(["ffmpeg", "-version"], capture_output=True, text=True)
            print(f"   Version Info: {res.stdout.splitlines()[0]}")
        except: pass
    else:
        print("❌ FFmpeg: NOT FOUND in PATH.")
        print("   Action: Please install ffmpeg (e.g., 'brew install ffmpeg' or 'conda install ffmpeg').")

    # 2. Library Checks
    libs = [
        ("mediapipe", "mp"),
        ("pandas", "pd"),
        ("numpy", "np"),
        ("cv2", "opencv-python"),
        ("scipy", "scipy"),
        ("plotly", "plotly")
    ]
    
    print("-" * 30)
    for lib, alias in libs:
        try:
            m = __import__(lib)
            ver = getattr(m, "__version__", "unknown")
            print(f"✅ {lib:<12} (v{ver})")
        
        except ImportError:
            print(f"❌ {lib:<12} NOT INSTALLED")

    print("-" * 30)
    print("Verification Complete!")

check_env()
````
